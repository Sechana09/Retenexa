```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
df = pd.read_csv("../data/processed/customer_churn_clean.csv")

df.head()
```

```python
##avg spend per person
df["AverageSpendPerOrder"] = (
    df["CashbackAmount"] /
    (df["OrderCount"] + 1)
)

df["AverageSpendPerOrder"].head()
```

```python
##engagement score
df["EngagementScore"] = (
    df["HourSpendOnApp"] +
    df["CouponUsed"] +
    df["OrderCount"]
)

df["EngagementScore"].head()
```

```python
##loyalty level
df["LoyaltyLevel"] = pd.cut(
    df["Tenure"],
    bins=[-1, 6, 12, 24, df["Tenure"].max()],
    labels=["New", "Regular", "Loyal", "VIP"]
)
```

```python
##high value customer
median_cashback = df["CashbackAmount"].median()

df["HighValueCustomer"] = (
    df["CashbackAmount"] >
    median_cashback
).astype(int)

df["HighValueCustomer"].value_counts()
```

```python
##customer activity
df["CustomerActivity"] = (
    df["OrderCount"] +
    df["CouponUsed"]
)

df["CustomerActivity"].head()
```

```python
##spending efficiency
df["CashbackPerOrder"] = (
    df["CashbackAmount"] /
    (df["OrderCount"] + 1)
)

df["CashbackPerOrder"].head()
```

```python
df.to_csv(
    "../data/features/customer_features.csv",
    index=False
)

print("Feature engineered dataset saved!")
```
